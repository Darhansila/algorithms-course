```python
df=pd.read_csv("data.csv")
```

```python
print(df.describe())
```

```python
import matplotlib.pyplot as plt
plt.scatter(df["age"], df["salary"])
plt.xlabel("age")
plt.ylabel("Salary")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
X = df[["age"]]
y = df["salary"]
model = LinearRegression()
model.fit(X, y)

print(model.coef_)
print(model.intercept_)
```

```python
import matplotlib.pyplot as plt

pred = model.predict(X)
plt.figure(figsize=(10, 6))

# Нақты мәндерді нүктелік график ретінде салу
plt.scatter(X, y, color="blue", alpha=0.7, label="Нақты мәндер")

# Регрессия сызығын салу
plt.plot(X, pred, color="red", linewidth=2, label="Регрессия сызығы")

# b_0 және b_1 мәндерін алу
b_0 = model.intercept_
b_1 = model.coef_[0]

# Формула мәтінін дайындау
formula_text = f"Formula: y = b0 + b1*x + e\nŷ = {b_0:.2f} + {b_1:.2f}*x"

# Графикке формуланы жазу (жоғары сол жақ бұрышқа)
plt.text(18, 480000, formula_text, fontsize=12, bbox=dict(facecolor='white', alpha=0.8, edgecolor='red'))

plt.xlabel("age")
plt.ylabel("Salary")
plt.title("Сызықтық регрессия моделі және формуласы")
plt.legend(loc="lower right")
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y, pred)

print("R² =", r2)
```
